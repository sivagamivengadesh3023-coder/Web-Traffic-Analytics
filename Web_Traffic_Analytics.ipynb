```python
df_numeric = df.copy()

for col in df_numeric.columns:
    converted = pd.to_numeric(df_numeric[col], errors='coerce')
    if converted.notna().sum() > 0:
        df_numeric[col] = converted

numeric_df = df_numeric.select_dtypes(include='number')

print("Numeric columns:", list(numeric_df.columns))
```

```python
categorical_cols = df.select_dtypes(include='object').columns
print(list(categorical_cols))
```

```python
for col in categorical_cols:
    print("\n", col)
    print(df[col].value_counts().head(10))
```

```python
if len(categorical_cols) > 0:
    col = categorical_cols[0]
    df[col].value_counts().head(10).plot(
        kind='bar', figsize=(10,5)
    )
    plt.title("Website Traffic Distribution")
    plt.xlabel(col)
    plt.ylabel("Count")
    plt.xticks(rotation=45)
    plt.show()
```

```python
if len(categorical_cols) > 0:
    col = categorical_cols[0]
    df[col].value_counts().head(5).plot(
        kind='pie', autopct='%1.1f%%',
        figsize=(7,7)
    )
    plt.title("Traffic Category Percentage")
    plt.ylabel("")
    plt.show()
```

```python
for col in categorical_cols:
    df[col].value_counts().to_csv(
        f"{col.replace(' ', '_')}_analysis.csv"
    )

print("Analysis files saved successfully!")
```

```python
for col in categorical_cols:
    print(col, ":", df[col].nunique())
```

```python
for col in categorical_cols:
    if df[col].notna().any():
        print("\nMost common in", col)
        print(df[col].mode().iloc[0])
```

```python
summary = []

for col in categorical_cols:
    counts = df[col].value_counts()
    if not counts.empty:
        summary.append({
            'Column': col,
            'Top Category': counts.index[0],
            'Count': counts.iloc[0]
        })

summary_df = pd.DataFrame(summary)
display(summary_df)
```

```python
if not summary_df.empty:
    plt.figure(figsize=(10,5))
    plt.bar(summary_df['Column'], summary_df['Count'])
    plt.title("Top Category Count by Column")
    plt.xticks(rotation=45)
    plt.ylabel("Count")
    plt.show()
```

```python
summary_df.to_csv("traffic_summary.csv", index=False)
print("Traffic summary exported successfully!")
```

```python
import plotly.express as px

fig1 = px.bar(
    summary_df,
    x='Column',
    y='Count',
    color='Column',
    title='Website Traffic Category Analysis'
)

fig1.show()
```

```python
if len(categorical_cols) > 0:
    col = categorical_cols[0]

    pie_data = df[col].value_counts().reset_index()
    pie_data.columns = ['Category', 'Count']

    fig2 = px.pie(
        pie_data.head(10),
        names='Category',
        values='Count',
        title=f'{col} Distribution'
    )

    fig2.show()
```

```python
fig1.write_html("traffic_dashboard.html")

print("Dashboard created successfully!")
```
